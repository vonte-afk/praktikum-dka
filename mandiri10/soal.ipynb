```python
#SOAL 1
import heapq

# 1 & 5. Representasi graph (adjacency list) dan nilai heuristik
graph = {
    'S': ['A', 'B'],
    'A': ['C', 'D'],
    'B': ['D', 'E'],
    'C': ['G'],
    'D': ['G'],
    'E': ['G'],
    'G': []
}

heuristic = {
    'S': 10, 'A': 5, 'B': 8, 'C': 3, 'D': 6, 'E': 9, 'G': 0
}

def greedy_bfs_soal1(graph, heuristic, start, goal):
    # Priority queue menyimpan tuple: (nilai_heuristik, node_sekarang, jalur_yang_ditempuh)
    # heapq secara default mengurutkan berdasarkan elemen pertama tuple (heuristik terkecil)
    open_list = []
    heapq.heappush(open_list, (heuristic[start], start, [start]))
    
    visited = set() # Mencegah siklus / pengulangan node
    nodes_visited_log = [] # Menyimpan urutan node yang dieksplorasi beserta heuristiknya

    while open_list:
        # Ambil node dengan nilai heuristik terkecil
        current_h, current_node, path = heapq.heappop(open_list)
        
        if current_node in visited:
            continue
            
        visited.add(current_node)
        nodes_visited_log.append((current_node, current_h))
        
        # Cek konfirmasi jika node tujuan berhasil ditemukan
        if current_node == goal:
            print("=== LOG EKSPLORASI NODE ===")
            for node, h in nodes_visited_log:
                print(f"Node dikunjungi: {node} (Heuristik: {h})")
            print("\n[KONFIRMASI] Node tujuan berhasil ditemukan!")
            print(f"Jalur lengkap: {' -> '.join(path)}")
            return True

        # Eksplorasi tetangga/adjacency list
        for neighbor in graph.get(current_node, []):
            if neighbor not in visited:
                # Masukkan ke priority queue berdasarkan nilai heuristik tetangga
                heapq.heappush(open_list, (heuristic[neighbor], neighbor, path + [neighbor]))
                
    print("\nNode tujuan tidak dapat ditemukan.")
    return False

# Jalankan pengujian program
greedy_bfs_soal1(graph, heuristic, 'S', 'G')
```
