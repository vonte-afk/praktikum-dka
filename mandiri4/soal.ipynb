```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df1 = pd.read_csv('soal1.csv')

# Membuat figure dengan 2 subplot (1 baris, 2 kolom)
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Subplot 1: Bar Plot (Menggunakan fungsi bawaan Pandas & Matplotlib)
df_counts = df1.groupby(['Pclass', 'Sex']).size().unstack(fill_value=0)
df_counts.plot(kind='bar', ax=axes[0], color=['pink', 'lightblue'], edgecolor='black')
axes[0].set_title('Jumlah Penumpang per Kelas (Pclass) & Gender')
axes[0].set_xlabel('Kelas Penumpang (Pclass)')
axes[0].set_ylabel('Jumlah Penumpang')
axes[0].set_xticklabels(df_counts.index, rotation=0)
axes[0].legend(title='Gender')

# Subplot 2: Box Plot (Memisahkan data usia manual)
age_not_survived = df1[df1['Survived'] == 0]['Age'].dropna()
age_survived = df1[df1['Survived'] == 1]['Age'].dropna()

axes[1].boxplot([age_not_survived, age_survived])
axes[1].set_title('Distribusi Usia berdasarkan Status Keselamatan')
axes[1].set_xlabel('Status Selamat')
axes[1].set_ylabel('Usia (Age)')
axes[1].set_xticklabels(['Tidak Selamat (0)', 'Selamat (1)'])

plt.tight_layout()
plt.show()
```

```python
#SOAL NO 2
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df2 = pd.read_csv('soal2.csv')

# Menampilkan statistik deskriptif untuk membantu analisis median
print("--- STATISTIK DESKRIPTIF NILAI ---")
print(df2[['math score', 'reading score']].describe())
print("-" * 40)

# Membuat figure baru untuk Soal 2
plt.figure(figsize=(14, 6))

# Subplot 1: Histogram Distribusi Math Score vs Reading Score
plt.subplot(1, 2, 1)
sns.histplot(df2['math score'], color='blue', label='Math Score', kde=True, alpha=0.5, bins=20)
sns.histplot(df2['reading score'], color='orange', label='Reading Score', kde=True, alpha=0.5, bins=20)
plt.title('Histogram Distribusi Nilai Siswa')
plt.xlabel('Nilai')
plt.ylabel('Frekuensi')
plt.legend()

# Subplot 2: Box Plot Membandingkan Math Score vs Reading Score
plt.subplot(1, 2, 2)
sns.boxplot(data=df2[['math score', 'reading score']], palette='Set3')
plt.title('Perbandingan Box Plot Nilai Siswa')
plt.ylabel('Nilai')
plt.xticklabels=['Math Score', 'Reading Score']

# Mengatur tata letak agar rapi
plt.tight_layout()

# Menampilkan plot
plt.show()
```
