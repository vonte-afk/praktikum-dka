```python
#SOAL 1
import networkx as nx

# 1. Membuat graph menggunakan nx.Graph() sesuai struktur organisasi
G1 = nx.Graph()

# Menambahkan edge berdasarkan relasi tree
edges_soal1 = [
    ('A', 'B'), ('A', 'C'),
    ('B', 'D'), ('B', 'E'),
    ('C', 'F'), ('C', 'G')
]
G1.add_edges_from(edges_soal1)

# 2 & 3. Implementasi DFS mulai dari node 'A' (Urutan DFS tanpa batas)
print("=== Urutan DFS Tanpa Batas ===")
dfs_unlimited = list(nx.dfs_edges(G1, source='A'))
for u, v in dfs_unlimited:
    print(f"Mengunjungi edge: {u} -> {v}")

print("-" * 30)

# 4. Batasi traversal hanya sampai depth 2
print("=== Urutan DFS dengan depth_limit=2 ===")
dfs_limited = list(nx.dfs_edges(G1, source='A', depth_limit=2))
for u, v in dfs_limited:
    print(f"Mengunjungi edge: {u} -> {v}")
```

```python
#SOAL 2
import networkx as nx

# 1. Membuat directed graph menggunakan nx.DiGraph()
G2 = nx.DiGraph()

# Menambahkan directed edges sesuai studi kasus
edges_soal2 = [
    ('A', 'B'), ('A', 'C'),
    ('B', 'D'), ('B', 'E'),
    ('C', 'E'), ('C', 'F'),
    ('E', 'F')
]
G2.add_edges_from(edges_soal2)

# Inisialisasi parameter pencarian
start_node = 'A'
goal_node = 'F'

# 2 & 3. Implementasi DFS mulai dari node 'A'
print("=== Menelusuri Jalur Navigasi (DFS) ===")

# Mendapatkan urutan edge dari DFS
dfs_edges = list(nx.dfs_edges(G2, source=start_node))

# Melacak urutan node yang dikunjungi (dimulai dari start_node)
visited_nodes = [start_node]
goal_found = False

for u, v in dfs_edges:
    # Catat edge yang dilewati
    print(f"Eksplorasi: {u} -> {v}")
    
    # Masukkan node tujuan ke daftar kunjungan jika belum ada
    if v not in visited_nodes:
        visited_nodes.append(v)
    
    # 5. (Opsional) Hentikan pencarian saat node tujuan ditemukan
    if v == goal_node:
        goal_found = True
        print(f"\n[Sinyal] Node tujuan '{goal_node}' TELAH DIALAMI / DITEMUKAN!")
        break

# Jika seluruh graf selesai dijelajahi dan baru dicek status akhirnya
if not goal_found:
    if goal_node in visited_nodes:
        goal_found = True

# 4. Tampilkan Hasil Akhir
print("\n=== Ringkasan Hasil ===")
print(f"Urutan Traversal Node DFS : {' -> '.join(visited_nodes)}")
print(f"Apakah Goal '{goal_node}' ditemukan? : {'YA' if goal_found else 'TIDAK'}")
```
